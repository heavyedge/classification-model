# Profiles by each type

```python
from heavyedge import ProfileData
import matplotlib.pyplot as plt

with ProfileData("./profiles.h5") as data:
    Ys, _, _ = data[:]

titles = [
    "Type 0",
    "Type 1a",
    "Type 1b",
    "Type 2a",
    "Type 2b",
    "Type 3a",
    "Type 3b",
]
for Y, title in zip(Ys, titles):
    plt.plot(Y)
    plt.title(title)
    plt.show()
```
